# Inheritance

## Example - 1

```python
class Person:
    def __init__(self,name):
        self.name = name

    def display(self):
        return f"your name is {self.name}"

class Student(Person):
    def __init__(self,name,marks):
        super().__init__(name)
        self.marks = marks

    def display(self):
        return f"your name is {self.name} and your mark is {self.marks}"

```

```python
P1 = Person("Vignesh")
P1.display()
```

```python
S1 = Student("Vignesh",93)
S1.display()
```

## Example - 2

```python
class Vehicle:
    def fuel_type(self):
        return "Fuel type not defined"

class Car(Vehicle):
    def fuel_type(self,fuel_type):
        self.fuel_type = fuel_type
        return f"Fuel type of Car is {self.fuel_type}"

class Bike(Vehicle):
    def fuel_type(self,fuel_type):
        self.fuel_type = fuel_type
        return f"Fuel type of Bike is {self.fuel_type}"
```

```python
v1 = Car()
print(v1.fuel_type("Diesel"))
```

```python
v2 = Bike()
print(v2.fuel_type("Petrol"))
```

# Polymorphism

## Example - 1

```python
class Shape:
    def area(self):
        return "Area not defined"

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return f"Circle Area: {3.14 * self.radius * self.radius}"

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print("Square Area:", self.side * self.side)

        
```

```python
ci = Circle(5)
sq = Square(4)
```

```python
print(ci.area())
print(sq.area())
```

## Example - 2

```python
class Employee:
    def work(self):
        pass

class Manager(Employee):
    def work(self):
        return "Managing Team"

class Developer(Employee):
    def work(self):
        return "Writing code"
```

```python
m = Manager()
d = Developer()
```

```python
print(m.work())
print(d.work())
```

# Abstraction

## Example - 1

```python
from abc import ABC, abstractmethod

class Bank(ABC):
    @abstractmethod
    def loan(self):
        pass

class SBI(Bank):
    def loan(self):
        return "Loan Approved"
```

```python
loan_status = SBI()
print(loan_status.loan())
```

## Example - 2

```python
from abc import ABC, abstractmethod

class Appliance:
    @abstractmethod
    def turn_on(self):
        pass
        
class Fan(Appliance):
    def turn_on(self):
        return "Fan spinning"


class WashingMachine(Appliance):
    def turn_on(self):
        return "Washing clothes"
```

```python
f = Fan()
w = WashingMachine()
```

```python
print(f.turn_on())
print(w.turn_on())
```
