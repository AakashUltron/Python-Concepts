```python
class laptop:

    charger_type = "C-Type"        # Class Variable
    
    def __init__(self):
        self.brand = ""
        self.price = 34

    def setPrice(self,price):
        self.price = price

    def getPrice(self):           
        print("Price:",self.price)

    @classmethod
    def changeChargerType(cls):
        cls.changeChargerType = "B-Type"
        print("Charger type changed to B")

samsung = laptop()
samsung.setPrice(20000)
samsung.getPrice()

laptop.changeChargerType()
```

```python
class laptop:

    charger_type = "C-Type"        # Class Variable
    
    def __init__(self):
        self.brand = ""
        self.price = 34

    def setPrice(self,price):
        self.price = price

    def getPrice(self):           
        print("Price:",self.price)

    @classmethod
    def changeChargerType(cls):
        cls.changeChargerType = "B-Type"
        print("Charger type changed to B")

    @staticmethod
    def info():
        print("this is laptop class")

samsung = laptop()
samsung.setPrice(20000)
samsung.getPrice()

laptop.changeChargerType()

samsung.info()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def display(self):
        print(self.name, self.salary, self.department)


m = Manager("AD", 100000, "CSE")
m.display()
```
