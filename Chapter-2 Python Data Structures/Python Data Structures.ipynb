## Types of Data Structures

### 1. List
* Duplicate Values
* One Dimensional
* Hetrogenous(can contain int,float,strings)
* Indexing & Sclicing is possible
* Mutubale(Can change values inside List)

```python
empty_cart = []
```

```python
empty_cart
```

```python
type(empty_cart)
```

```python
items_to_buy = ["Apple","Orange","Pomegranate","Apple",6,98,90.8]
items_to_buy
```

```python
type(items_to_buy)
```

### Note:
#### During Indexing:
* Always positive indexing starts from 0 (starts from left to right).           
* Always Negative indexing starts from -1 (starts from right to left).

#### During Sclicing:
* for positive indexing:
    it starts from 0 and ending should be the next index number to get exact Value.
* for negative indexing:
    it Starts from -1 (Moves right → left)

```python
items_to_buy[0]
```

```python
items_to_buy[-3]
```

```python
items_to_buy[2:6]
```

```python
items_to_buy[-6:-1]
```

```python
items_to_buy[2] = "StarFuit"
items_to_buy
```

### 2. Sets
* it Doesn't allow Duplicate Values
* One Dimensional
* Hetrogenous(can contain int,float,strings)
* Indexing & Sclicing is not possible
* Mutubale(Can change values inside Sets)

```python
empty_set = set()
type(empty_set)
```

```python
items_to_buy_2 = {"Apple","Orange","Pomegranate","Apple",6,98,90.8}
items_to_buy_2
```

```python
items_to_buy_2.add("Potato")
```

```python
items_to_buy_2
```

```python
items_to_buy_2.discard("Apple")
```

```python
items_to_buy_2
```

### 3. Tuples
* it allow Duplicate Values
* One Dimensional
* Hetrogenous(can contain int,float,strings)
* Indexing & Sclicing is possible
* Immutubale(Cannot change values inside Tuple)

```python
empty_tuple = ()
type(empty_tuple)
```

```python
items_to_buy_3 = ("Apple","Orange","Pomegranate","Apple",6,98,90.8)
items_to_buy_3
```

```python
items_to_buy_3[0:3]
```

```python
items_to_buy_3.count("Apple")
```

```python
items_to_buy_3.index("Apple")
```

### 4. Dictionary
It is a key value pair
* Don't create duplicate keys, it allow Duplicate Values but duplicate value is updated to original value
* Two Dimensional
* Hetrogenous(can contain int,float,strings)
* Indexing & Sclicing is not possible
* Mutable(Can change values inside Dictionary)

```python
empty_dictionary = {}
type(empty_dictionary)
```

```python
student_marks = {"Maths":76,
                "English":93,
                "Science":98}
```

```python
student_marks
```

```python
items_to_buy_4 = {"Apple":5,"Orange":3,"Pomegranate":1,"Apple":4} # Refer the 1st point in Dictionary
items_to_buy_4
```

```python
items_to_buy_4.clear()
items_to_buy_4
```

```python
items_to_buy_5 = {"Apple":5,"Orange":3,"Pomegranate":1}
items_to_buy_5.keys()
```

```python
items_to_buy_5.values()
```
