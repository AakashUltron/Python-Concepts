## Functions & Loops

### Positional Arguments

```python
def get_details(name,place,expected_price):
    print("Client Name:", name)
    print("Place:", place)
    print("Expected Price:", expected_price)
```

```python
get_details(place = "Chennai",expected_price = 98,name = "Ultron")
```

### Default parameters & Arguments

```python
def get_details(name,expected_price,place = "Chennai"):
    print("Client Name:", name)
    print("Place:", place)
    print("Expected Price:", expected_price)
```

```python
get_details(expected_price = 98,name = "Ultron")
```

### Variable - length Arguments

```python
def get_details(name,place,*expected_price):
    print("Client Name:", name)
    print("Place:", place)
    print("Expected Price:", expected_price)
```

```python
get_details("Ultron","Chennai", 98,91,90.8)
```

### Keyword Arguments

```python
def get_details(name,place,**expected_price):
    print("Client Name:", name)
    print("Place:", place)
    print("Expected Price:", expected_price)
```

```python
get_details("Ultron","Chennai", ask_1 = 98,ask_2 = 91,ask_3 = 90.8)
```

## Control the Flow of the Program

```python
user_password = input("New Password:")

if len(user_password) < 5:
    print("Weak Password")
elif 5 <= len(user_password) <= 10:
    print("Moderate Password")
else:
    print("Strong Password")
```

### Try it Yourself

```python
user_input = int(input("Enter your number:"))

for i in range (1,user_input+1):
    if user_input%i == 0:
        print(i)
```

```python
user_input = int(input("Enter your number:"))

result = []
for i in range (1,user_input+1):
    if user_input%i == 0:
        result.append(i)

print(f"The Factors of {user_input} is {result}")
```
