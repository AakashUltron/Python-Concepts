```python
import pandas as pd
```

## 1) Groupby Function

```python
insurance_data = pd.read_csv("insurance.csv")
insurance_data
```

### 1.1 Perform Initial Analysis

```python
insurance_data.isnull().sum()
```

```python
insurance_data.dtypes
```

```python
insurance_data.describe()
```

### 1.2 Pull Up the Insurance Charges based on Region

```python
insurance_data["region"].unique()
```

```python
insurance_data["region"].nunique()
```

```python
insurance_data.groupby(by = "region")["charges"].mean().round(2).sort_values(ascending = False)
```

### 1.3 Pull up the average Insurance charges based on region and sex

### Discrete and Continous Data:
- Discrete - Just Countable/Cannot be Segmented.
- Continous - Measurable/Have units/Can also be Segmented.

```python
insurance_data.groupby(by = ["region","sex"])["charges"].max().round(2).sort_values(ascending = False)
```

## 2) Pivot Table

```python
pd.pivot_table(data = insurance_data, values = "charges", index = "region", columns = "sex").round()
```

```python
pd.pivot_table(data = insurance_data, values = "charges", index = "region", columns = ["sex","children"]).round()
```

## 3) Aggregate

### 3.1 Find out the average insurance charges based on region

```python
insurance_data.groupby(by = "region")["charges"].aggregate(func = ["mean","min","max","sum"]).round()
```

```python
insurance_data.groupby(by = ["region","sex"])["charges"].aggregate(func = ["mean","min","max","sum"]).round() 
```

## 4) Cross Tab

### 4.1 How many Smokers and Non Smokers based on region?

```python
pd.crosstab(index = insurance_data["region"], columns = insurance_data["smoker"])
```

```python
pd.crosstab(index = insurance_data["region"], columns = insurance_data["smoker"],margins = True)
```

## 6) Apply Function

### 6.1 Create a new column in the same DataFrame with the name "Eligiblity Criteria" where if age is > 20, say Eligible else not.

```python
def get_age(x):
    if x > 20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
insurance_data["age"].apply(get_age)
```

```python
insurance_data["Eligiblity Criteria"] = insurance_data["age"].apply(get_age)
insurance_data
```

## 7) Lambda Function
- Anonymous Function/nameless Function

```python
insurance_data["Eligiblity Criteria 2"] = insurance_data["age"].apply(lambda x:1 if x>20 else 0)
insurance_data
```

```python
insurance_data["Eligiblity Criteria 3"] = insurance_data["age"].apply(lambda x:"Eligible" if x>20 else "Not Eligible")
insurance_data
```

## 5) Concat Function

```python
sales_data_2017 = pd.read_csv("Sales Transactions-2017.csv")
sales_data_2018 = pd.read_csv("Sales Transactions-2018.csv")
sales_data_2019 = pd.read_csv("Sales Transactions-2019.csv")
```

```python
sales_full_data = pd.concat(objs = [sales_data_2017,sales_data_2018,sales_data_2019])
sales_full_data
```
