## Stone Paper Scissors Game

```python
import random

print("Welcome to Stone Paper Scissors Game")

choices = ["stone", "paper", "scissors"]

while True:
    user_choice = input("Enter your choice (stone/paper/scissors): ").lower()

    if user_choice not in choices:
        print("Invalid choice! Try again.")
        continue

    computer_move = random.choice(choices)

    print("You chose:", user_choice)
    print("Computer chose:", computer_move)

    if user_choice == computer_move:
        print("It's a Tie!")

    elif (
        (user_choice == "stone" and computer_move == "scissors") or
        (user_choice == "paper" and computer_move == "stone") or
        (user_choice == "scissors" and computer_move == "paper")
    ):
        print("You Win!")

    else:
        print("Computer Wins!")

    play_again = input("Do you want to play again? (yes/no): ").lower()
    if play_again != "yes":
        print("Thanks for playing!")
        break
```

```python
import random

print("Welcome to Word Guessing Game")

Words = ["python", "machine", "learning", "bridge"]

Word = random.choice(Words)

hidden_word = []
attempts = 6
guessed_letters = []

for letter in Word:
    hidden_word.append("_")

while attempts > 0:
    print("\nWord:", " ".join(hidden_word))
    print("Attempts left:", attempts)
    print("Guessed letters:", ", ".join(guessed_letters))

    letter = input("Enter a letter: ").lower()

    if len(letter) != 1 or not letter.isalpha():
        print("Enter only one valid letter")
        continue

    if letter in guessed_letters:
        print("Already guessed")
        continue

    guessed_letters.append(letter)

    if letter in Word:
        for i in range(len(Word)):
            if Word[i] == letter:
                hidden_word[i] = letter
    else:
        attempts -= 1
        print("Wrong guess!")

    if "_" not in hidden_word:
        print("You Win!")
        print("The word was:", Word)
        break

if attempts == 0:
    print("You Lost!")
    print("The correct word was:", Word)
```
