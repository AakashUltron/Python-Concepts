# Number Guessing Game

## Number Guessing game using While Loop

```python
secret_number = 13

while True:
    user_input = int(input("Enter Your Guess:"))

    if secret_number == user_input:
        print("Your Guess is Right! Well Done...")
        break;
    else:
        print("Your Guess is Wrong! Try Again...")
```

## Number Guessing game using For Loop

```python
secret_number = 13

for i in range(0,7):
    user_input = int(input("Enter Your Guess:"))

    if (secret_number == user_input):
        print("Your Guess is Right! Well Done...")
        break;
    else:
        print("Your Guess is Wrong! Try Again...")
```

## Number Guessing game using Nested If else


```python
print("Welcome to Number Guessing Game!")
print("Enter the Number between 1 to 100!")


secret_number = 13

for i in range(1,7):
    user_input = int(input("Enter Your Guess:"))

    if (user_input < 1 or user_input > 100):
        print("Invalid number ! please enter number between 1 to 100 only")
    elif (user_input > secret_number):
        print("Your number is higher!")
    elif (user_input < secret_number):
        print("Your number is Lesser!")
    else:
        print("Your Guess is Right! Well Done...")
        break;

if (secret_number == user_input):
    print(f"You Have Guessed the number {secret_number} in {i} attempts")
else:
    print("Game Over! All your Trails has Finished 😢")
```

## Number Guessing game to handle Invalid Characters

```python
import random

print("Welcome to Number Guessing Game!")
print("Enter the Number between 1 to 100!")


secret_number = random.randint(1,100)

for i in range(1,7):
    try:
        user_input = int(input("Enter Your Guess:"))
    
        if (user_input < 1 or user_input > 100):
            print("Invalid number ! please enter number between 1 to 100 only")
        elif (user_input > secret_number):
            print("Your number is higher!")
        elif (user_input < secret_number):
            print("Your number is Lesser!")
        else:
            print("Your Guess is Right! Well Done...")
            break;
    except ValueError:
        print("Invalid Charcters, Not Recognised. Please enter number between 1 to 100 only")

if (secret_number == user_input):
    print(f"You Have Guessed the number {secret_number} in {i} attempts")
else:
    print("Game Over! All your Trails has Finished 😢")
```
