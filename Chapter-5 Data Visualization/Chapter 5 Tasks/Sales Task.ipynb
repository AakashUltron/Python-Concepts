# Univariate Analysis

```python
plt.hist(superstore["Quantity"])
plt.xlabel("Quantity of items")
plt.ylabel("counts")
```

```python
sns.countplot(superstore, x="Category")
```

```python
sns.countplot(superstore, x = "Quantity")
```

```python
sns.countplot(superstore, x = "Sub-Category", hue = "Category")
plt.xticks(rotation = 90)
plt.show()
```

```python
plt.boxplot(superstore["Quantity"])
```

```python
superstore["Segment"].value_counts()
```

```python
plt.pie(superstore["Segment"].value_counts(), labels = ["Consumer","Corporate","Home Office"], autopct = '%1.0f%%')
plt.show()
```

```python
superstore["Ship Mode"].value_counts()
```

```python
plt.pie(superstore["Ship Mode"].value_counts(),labels = ["Standard Class","Second Class","First Class","Same Day"], autopct = '%1.01f%%')
plt.show()
```

```python
superstore["Category"].value_counts()
```

```python
plt.pie(superstore["Category"].value_counts(),labels = ["Office Supplies","Furniture","Technology"], autopct = '%1.0f%%')
plt.show()
```

# Bivariate Analysis

```python
sns.barplot(superstore, x = "Region", y = "Sales")
```

```python
sns.barplot(superstore, x = "Quantity", y = "Sales")
```

```python
plt.bar(superstore["Quantity"],height = superstore["Sales"])
```

```python
sns.scatterplot(superstore, x = "Sales", y = "Profit", hue = "Category")
```

```python
sns.scatterplot(superstore, x = "Quantity", y = "Sales", hue = "Segment")
```

```python
sns.boxplot(x="Category", y="Sales", data=superstore)
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x="Category",
    y=np.log1p(superstore["Sales"]),
    data=superstore
)

plt.ylabel("Sales")
plt.title("Category vs Sales")
plt.show()
```

# Multivariate Analysis

```python
sns.pairplot(superstore)
```

```python
Superstore_corr = superstore[["Sales","Quantity","Discount","Profit"]].corr()
Superstore_corr
```

```python
sns.heatmap(Superstore_corr, annot=True)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    Superstore_corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Analysis")
plt.show()
```
