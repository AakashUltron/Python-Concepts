```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'Other countries' # Represent only large countries
fig = px.pie(df, values='pop', names='country', title='Population of European continent')
fig.show()
```

## Note:
1. Univariate charts can include more than one feature.
2. Bivariate charts can include more than two feature.

```python
penguins = sns.load_dataset("penguins")
sns.histplot(data = penguins, x = "flipper_length_mm", kde = True, bins = 20, color = "skyblue")
plt.title("Flipper Length Distribution with KDE")
plt.show()
```
