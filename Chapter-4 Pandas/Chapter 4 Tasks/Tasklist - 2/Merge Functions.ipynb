```python
import pandas as pd
```

```python
ordered_items = pd.read_csv("orders.csv")
ordered_items
```

```python
customers_details = pd.read_csv("customers.csv")
customers_details
```

```python
pd.merge(customers_details,ordered_items)
```

```python
pd.merge(customers_details,ordered_items, on='customer_id', how='inner')
```

```python
pd.merge(customers_details,ordered_items, on='customer_id', how='left')
```

```python
pd.merge(customers_details,ordered_items, on='customer_id', how='right')
```

```python
pd.merge(customers_details,ordered_items, on='customer_id', how='outer')
```

```python
pd.merge(customers_details,ordered_items, how='cross')
```
