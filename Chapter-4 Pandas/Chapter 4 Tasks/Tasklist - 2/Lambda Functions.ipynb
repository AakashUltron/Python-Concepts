# 1) Map

## Problem 1 - Create a new list where each number is cube (x³)

```python
nums = [2, 4, 6, 8]
cube_values = list(map(lambda x: x*3, nums))
print(cube_values)
```

## Problem 2 - Convert each name to: 
- first letter capital
- rest lowercase

```python
names = ["aakash", "regwin", "rohit"]
updates_case = list(map(lambda x: x[0].upper()+x[1:].lower(),names))
print(updates_case)
```

```python
names = ["aakash", "regwin", "rohit"]
updated_cases = list(map(lambda x: x.capitalize(), names))
print(updates_cases)
```

# 2) Filter

## Problem 1 - Filter numbers that are divisible by 5 AND greater than 10

```python
numerics = [5, 10, 15, 20, 25]
divised_nums = list(filter(lambda x: x % 5 == 0 and x > 10 ,numerics))
print(divised_nums)
```

## Problem 2 - Filter words with length > 3

```python
words = ["apple", "bat", "car", "elephant", "dog"]
filtered_words = list(filter(lambda x: len(x) > 3 , words))
print(filtered_words)
```

# 3) Reduce

## Problem 1 - Find the product of all numbers

```python
from functools import reduce
```

```python
numbers = [2, 3, 4]
product_values = reduce(lambda x,y: x*y , numbers)
print(product_values)
```

## Problem 2 - Find the maximum number using reduce

```python
num = [10, 5, 8, 20]
maximum_number = reduce(lambda x,y: x and y , num)
print(maximum_number)
```
